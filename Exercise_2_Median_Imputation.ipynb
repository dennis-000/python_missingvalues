```python
import pandas as pd
import numpy as np

# Reload for fresh start
df = pd.read_csv('sensor_log.csv')
numeric_cols = df.select_dtypes(include='number').columns

df_mean = df.copy()
for col in numeric_cols:
    col_mean = df_mean[col].mean()
    df_mean[col] = df_mean[col].fillna(col_mean)

df_median = df.copy()
for col in numeric_cols:
    col_median = df_median[col].median()
    df_median[col] = df_median[col].fillna(col_median)

print("df_mean (from notebook):")
print(df_mean[numeric_cols].head())
print("\ndf_median:")
print(df_median[numeric_cols].head())

print("\nMeans vs. Medians per column:")
for col in numeric_cols:
    print(f"{col}: Mean={df[col].mean():.2f}, Median={df[col].median():.2f}")
```
