```python
import pandas as pd

# Load the uploaded dataset
df = pd.read_csv("sensor_log.csv")
```

```python
# Display missing value counts
# To identify the missing values easily (Nulls or Nan)
df.isnull().sum()
```

```python
# Convert timestamp to datetime and set as index
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.set_index('timestamp', inplace=True)

# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())

df  # Display the prepared DataFrame
```

```python
# 1. Forward Fill: Propagate previous valid value forward
df_ffill = df.copy()
df_ffill = df_ffill.ffill()

print("Forward Fill DataFrame:")
print(df_ffill)

print("\nSummary Statistics:")
print(df_ffill.describe())
```

```python
# 2. Backfill: Propagate next valid value backward
df_bfill = df.copy()
df_bfill = df_bfill.bfill()

print("Backfill DataFrame:")
print(df_bfill)

print("\nSummary Statistics:")
print(df_bfill.describe())
```

```python
# 3. Interpolation: Linear interpolation weighted by time intervals
df_interp = df.copy()
df_interp = df_interp.interpolate(method='time')

print("Interpolation DataFrame:")
print(df_interp)

print("\nSummary Statistics:")
print(df_interp.describe())
```

```python
# 4. Mean Fill: Replace with column-specific means
df_mean = df.copy()
df_mean = df_mean.fillna(df_mean.mean())

print("Mean Fill DataFrame:")
print(df_mean)

print("\nSummary Statistics:")
print(df_mean.describe())
```

```python
df_ffill.to_csv('/content/sensor_log_ffill.csv')
df_bfill.to_csv('/content/sensor_log_bfill.csv')
df_interp.to_csv('/content/sensor_log_interp.csv')
df_mean.to_csv('/content/sensor_log_mean.csv')

print("Filled DataFrames ready: df_ffill, df_bfill, df_interp, df_mean")
```
